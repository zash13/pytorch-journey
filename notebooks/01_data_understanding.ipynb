```
import re
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "axes.titleweight": "bold",
})

def safe_name(s: str) -> str:
    s = s.strip()
    s = re.sub(r"[^\w\-]+", "_", s)
    return s.strip("_") or "unnamed"

MAX_POINTS = 5000

for column in incpect_column_names:
    if column == target_column:
        continue

    sub = df[[column, target_column]].dropna()
    r = sub[column].corr(sub[target_column])
    n_total = len(sub)

    if n_total > MAX_POINTS:
        sub_plot = sub.sample(MAX_POINTS, random_state=42)
    else:
        sub_plot = sub
    n_plot = len(sub_plot)

    fig, ax = plt.subplots(figsize=(12, 7))   # wider & taller, more open

    sns.regplot(
        data=sub_plot, x=target_column, y=column, 
        scatter_kws={
            "facecolor": "none",
            "edgecolor": "steelblue",
            "s": 40,                
            "alpha": 0.45,     
            "linewidths": 1.1,
        },
        #    line_kws={"color": "crimson", "linewidth": 2.0},
        ci=95,                      
    )

    subtitle = f"r = {r:.3f}    n = {n_total}"
    if n_plot < n_total:
        subtitle += f"    (plotted: {n_plot} sampled)"

    ax.set_title(f"{column}  vs  {target_column}\n{subtitle}", pad=14)
    ax.set_xlabel(target_column, labelpad=10)
    ax.set_ylabel(column, labelpad=10)
    ax.grid(True, linestyle="--", alpha=0.35)
    ax.set_axisbelow(True)          # grid behind bubbles

    x0, x1 = ax.get_xlim()
    y0, y1 = ax.get_ylim()
    ax.set_xlim(x0 - 0.03 * (x1 - x0), x1 + 0.03 * (x1 - x0))
    ax.set_ylim(y0 - 0.03 * (y1 - y0), y1 + 0.03 * (y1 - y0))

    fname = f"reports/plots/corr_{safe_name(column)}_vs_{safe_name(target_column)}.png"
    plt.savefig(fname, dpi=150, bbox_inches="tight")
    plt.close(fig)

    print(f"{column:20s}  r = {r: .3f}   n = {n_total}   ->  {fname}")
```
